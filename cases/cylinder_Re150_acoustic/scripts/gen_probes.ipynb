```python
import numpy as np
import pandas as pd

# ==================== 参数设置 ====================
diameter = 1.0          # 圆柱直径 (1m)
# radius = diameter / 2.0 # 圆柱半径
radius = 75 * diameter # acoustic probes
# num_points_per_ring = 180  # 每圈的测点数量（每2度一个点）
num_points_per_ring = int(360/5)  # 每圈的测点数量（每2度一个点）
# y_sections = [-0.5, 0.0, 0.5]  # 展向(y轴)监测切面位置
y_sections = [0.0]  # 展向(y轴)监测切面位置
eps = 0

# ==================== 坐标计算 ====================
point_list = []

for y in y_sections:
    for i in range(num_points_per_ring):
        # 几何夹角 theta（从 0 到 2*pi）
        theta = 2 * np.pi * i / num_points_per_ring
        
        # 圆柱中心在 (0,0,0)，展向为 y 轴
        # 截面圆在 x-z 平面上
        x = (radius + eps) * np.cos(theta)
        z = (radius + eps) * np.sin(theta)
        
        # 转换为角度制 (0° ~ 360°)
        theta_deg = np.degrees(theta)
        
        point_list.append({
            # 'Point_ID': f"Y_{y:.1f}_P{i:03d}",
            'X': round(x, 6),
            'Y': round(y, 6),
            'Z': round(z, 6),
            # 'Theta_deg': round(theta_deg, 2)
        })

# ==================== 导出数据 ====================
df = pd.DataFrame(point_list)
# output_filename = "cylinder_measurement_points.csv"
output_filename = "cylinderRe150_acoustic_probes.csv"
df.to_csv(output_filename, index=False, encoding='utf-8')

print(f"成功生成测点文件：{output_filename}，共包含 {len(df)} 个测点。")
```
