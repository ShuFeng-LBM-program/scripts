```python
import numpy as np
numPt = 41
r = 1.905
angleRange = np.linspace(-45, 45, numPt, True)
print(angleRange)
radRange = np.deg2rad(angleRange)

xVec = -r * np.cos(radRange)
zVec = -r * np.sin(radRange)
yVec = np.full(numPt, 0.0)

data = np.column_stack([xVec, yVec, zVec])
np.savetxt("DJI_microphones_m.csv", data, delimiter=",", header="x,y,z", comments="", fmt="%.6f")
```
